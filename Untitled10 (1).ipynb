```python
# 1. Simple Function
from datetime import datetime

def welcome_user(student_name):
    live_time = datetime.now().strftime("%I:%M:%S %p")

    print("Welcome", student_name)
    print("Time Now:", live_time)

welcome_user("Sara")


# 2. Function with Multiple Parameters

def calculate_average(value_x, value_y, value_z):
    average_total = (value_x + value_y + value_z) / 3
    return average_total

print("Average Value:", calculate_average(12, 18, 24))


# 3. Default Argument

def calculate_discount(product_cost, discount_value=10):
    saved_amount = (product_cost * discount_value) / 100
    final_bill = product_cost - saved_amount
    return final_bill

print("Discount Price:", calculate_discount(5000))
print("Discount Price:", calculate_discount(5000, 25))


# 4. Keyword Arguments

def print_info(person_name, person_age, person_city):
    print("Student Name:", person_name)
    print("Student Age:", person_age)
    print("Student City:", person_city)

print_info(person_age=19, person_city="Karachi", person_name="Bilal")


# 5. Variable-Length Arguments

def find_max(*number_group):
    return max(number_group)

print("Largest Number:", find_max(11, 45, 7, 89, 32))


# 6. Recursive Function

def factorial_number(input_num):
    if input_num == 1:
        return 1

    return input_num * factorial_number(input_num - 1)

print("Factorial Answer:", factorial_number(6))


# 7. Function as an Argument

def cube_value(check_num):
    return check_num * check_num * check_num

def apply_function(work_func, list_data):
    updated_values = []

    for current_item in list_data:
        updated_values.append(work_func(current_item))

    return updated_values

main_numbers = [2, 4, 6, 8]

print("New List:", apply_function(cube_value, main_numbers))
```
