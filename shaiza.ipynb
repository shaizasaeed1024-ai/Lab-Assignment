```python
inv={}
def add(n,q,p): inv[n]={"q":q,"p":p}
def total(): return sum(i["q"]*i["p"] for i in inv.values())
def low(t=5): return {k:v for k,v in inv.items() if v["q"]<t}

add("laptop",2,1000)
add("phone",3,500)

print(total())
print(low())
```

```python
s={"Ali":{"m":80},"Sara":{"m":90}}
avg={k:sum(v.values())/len(v) for k,v in s.items()}

print(avg)
print(max(s,key=lambda x:s[x]["m"]))
print(sorted(avg,key=avg.get,reverse=True))
```

```python
store={"a":{"p":10,"q":5}}
cart={}

def add(x,n): cart[x]=cart.get(x,0)+n
def total(d=0,t=0):
    return (sum(store[i]["p"]*q for i,q in cart.items())*(1-d))*(1+t)

add("a",2)

print(cart)
print(total(0.1,0.05))
```
